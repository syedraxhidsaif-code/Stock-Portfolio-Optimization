# 📊 Stock Portfolio Optimization & Risk Analysis

## Notebook 02: Data Cleaning & Quality Checks

### Objective
Validate the downloaded stock market data, check for missing values and duplicates, prepare a clean Adjusted Close price dataset, and export it for portfolio analysis.

### Tasks
- Load raw stock data
- Check data quality
- Handle missing values
- Check duplicate rows
- Verify datetime index
- Extract Adjusted Close prices
- Export cleaned dataset

```python
import pandas as pd 
import numpy as np
```

Load Rwa Stock Data 

```python
data = pd.read_csv(
    "../data/raw/stock_data.csv",
    header=[0,1],
    index_col=0,
    parse_dates=True
)

data.head()
```

Basic Information

```python
print("="*50)
print("Dataset Shape")
print(data.shape)

print("="*50)
print("Dataset Information")
data.info()
```

Missing Value

```python
missing_values = data.isnull().sum()

print(missing_values)
```

```python
print("Total Missing Values:", missing_values.sum())
```

Duplicates Rwa

```python
duplicates = data.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

Data Types


```python
print(data.dtypes)
```

Verify Datetime Index

```python
print(type(data.index))

print(data.index)
```

Extract Adjusted Close Prices

```python
adj_close = data.xs("Adj Close", level=1, axis=1)

adj_close.head()
```

Rename Columns

```python
adj_close.columns = [
    col.replace(".NS", "")
    for col in adj_close.columns
]

adj_close.head()
```

```python
adj_close.info()

adj_close.describe()
```

```python
adj_close.to_csv("../data/processed/adjusted_close.csv")

print("Clean dataset exported successfully!")
```
