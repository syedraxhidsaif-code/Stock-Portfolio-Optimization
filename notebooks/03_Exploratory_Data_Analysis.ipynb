# Step 20 – Calculate Compound Annual Growth Rate (CAGR)

## Objective

Compound Annual Growth Rate (CAGR) measures the average annual rate at which an investment grows over a specified period, assuming returns are reinvested.

Unlike simple average returns, CAGR accounts for the effects of compounding and provides a more accurate measure of long-term investment performance.

Portfolio managers and investment analysts frequently use CAGR to compare investments with different return patterns.

```python
# ==========================================
# Calculate CAGR
# ==========================================

# Beginning prices
beginning_price = adj_close.iloc[0]

# Ending prices
ending_price = adj_close.iloc[-1]

# Number of years
years = (adj_close.index[-1] - adj_close.index[0]).days / 365.25

# CAGR Calculation
cagr = ((ending_price / beginning_price) ** (1 / years) - 1) * 100

# Convert to DataFrame
cagr_df = pd.DataFrame({
    "CAGR (%)": cagr.round(2)
})

# Sort from highest to lowest
cagr_df = cagr_df.sort_values("CAGR (%)", ascending=False)

cagr_df
```

# Observation

The CAGR table ranks stocks based on their average annual growth over the investment period.

## Business Insights

- Higher CAGR indicates stronger long-term performance.
- CAGR smooths yearly fluctuations, making comparisons more meaningful.
- Investors often use CAGR to identify companies with consistent long-term growth.
- It is a preferred metric in equity research and portfolio management.

# Step 21 – Visualize CAGR

## Objective

This helps quickly identify the best and weakest long-term performers.

```python
# ==========================================
# Plot CAGR
# ==========================================

plt.figure(figsize=(12,6))

plt.bar(cagr_df.index, cagr_df["CAGR (%)"])

plt.title("Compound Annual Growth Rate (CAGR)", fontsize=16)

plt.xlabel("Stocks")

plt.ylabel("CAGR (%)")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```

# Step 30 – Calculate Downside Risk

## Objective

Downside Risk measures the volatility of negative returns only.

Unlike standard deviation, it focuses exclusively on losses, making it particularly useful for evaluating downside exposure.

This metric is commonly used in portfolio management and risk assessment.

```python
# ==========================================
# Calculate Annualized Downside Risk
# ==========================================

# Keep only negative returns
downside_returns = daily_returns.where(daily_returns < 0, 0)

# Annualized downside risk
downside_risk = np.sqrt((downside_returns ** 2).mean()) * np.sqrt(252)

# Create DataFrame
downside_risk_df = pd.DataFrame({
    "Annualized Downside Risk": downside_risk
})

# Sort highest to lowest
downside_risk_df = downside_risk_df.sort_values(
    by="Annualized Downside Risk",
    ascending=False
)

downside_risk_df
```

# Observation

The table ranks stocks according to their annualized downside risk.

## Business Insights

- Higher downside risk indicates larger and more frequent negative returns.
- Lower downside risk suggests stronger downside protection.
- Conservative investors often prefer assets with lower downside risk.

```python
plt.figure(figsize=(12,6))

plt.bar(
    downside_risk_df.index,
    downside_risk_df["Annualized Downside Risk"]
)

plt.title("Annualized Downside Risk")

plt.xlabel("Stocks")

plt.ylabel("Downside Risk")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```

# Step 32 – Calculate Maximum Drawdown

## Objective

Maximum Drawdown (MDD) measures the largest historical decline from a previous peak in cumulative returns.

It helps investors understand the worst-case historical loss experienced during the investment period.

```python
# ==========================================
# Calculate Maximum Drawdown
# ==========================================

# Wealth Index
wealth_index = (1 + daily_returns).cumprod()

# Running Maximum
running_max = wealth_index.cummax()

# Drawdown
drawdown = (wealth_index - running_max) / running_max

# Maximum Drawdown
max_drawdown = drawdown.min()

max_drawdown_df = pd.DataFrame({
    "Maximum Drawdown": max_drawdown
})

max_drawdown_df = max_drawdown_df.sort_values(
    by="Maximum Drawdown"
)

max_drawdown_df
```

# Observation

Maximum Drawdown measures the worst historical decline for each stock.

## Business Insights

- Larger negative values indicate deeper historical losses.
- Lower drawdowns suggest stronger capital preservation.
- Portfolio managers use MDD to evaluate downside exposure and recovery potential.

```python
plt.figure(figsize=(12,6))

plt.bar(
    max_drawdown_df.index,
    max_drawdown_df["Maximum Drawdown"]
)

plt.title("Maximum Drawdown")

plt.xlabel("Stocks")

plt.ylabel("Drawdown")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```

# Step 34 – Calculate Value at Risk (VaR)

## Objective

Value at Risk (VaR) estimates the maximum expected loss over a specified period at a chosen confidence level.

In this project, we calculate the historical 95% VaR using daily returns.

VaR is widely used in:

- Investment Banking
- Portfolio Management
- Risk Management
- Regulatory Reporting

```python
# ==========================================
# Calculate Historical Value at Risk (95%)
# ==========================================

confidence_level = 0.05

var_95 = daily_returns.quantile(confidence_level)

var_df = pd.DataFrame({
    "95% VaR": var_95
})

# Sort from highest risk to lowest
var_df = var_df.sort_values(
    by="95% VaR"
)

var_df
```

# Observation

The table presents the 95% historical Value at Risk (VaR) for each stock.

## Business Insights

- More negative VaR values indicate greater downside risk.
- Less negative VaR values suggest relatively lower potential losses.
- VaR estimates the loss threshold that is expected to be exceeded only 5% of the time.

```python
plt.figure(figsize=(12,6))

plt.bar(
    var_df.index,
    var_df["95% VaR"]
)

plt.title("95% Historical Value at Risk")

plt.xlabel("Stocks")

plt.ylabel("Daily Return")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```

# Step 36 – Calculate Conditional Value at Risk (CVaR)

## Objective

Conditional Value at Risk (CVaR) measures the average loss that occurs beyond the Value at Risk threshold.

CVaR provides a better understanding of tail risk and is commonly used in advanced portfolio risk management.

```python
# ==========================================
# Calculate Historical CVaR (95%)
# ==========================================

cvar = daily_returns.apply(
    lambda x: x[x <= x.quantile(0.05)].mean()
)

cvar_df = pd.DataFrame({
    "95% CVaR": cvar
})

cvar_df = cvar_df.sort_values(
    by="95% CVaR"
)

cvar_df
```

# Observation

The table reports the average loss during the worst 5% of trading days.

## Business Insights

- More negative CVaR values indicate greater exposure to extreme losses.
- CVaR complements VaR by measuring the severity of losses beyond the VaR threshold.
- Portfolio managers often rely on CVaR when evaluating tail risk.

```python
plt.figure(figsize=(12,6))

plt.bar(
    cvar_df.index,
    cvar_df["95% CVaR"]
)

plt.title("95% Conditional Value at Risk (CVaR)")

plt.xlabel("Stocks")

plt.ylabel("Daily Return")

plt.grid(axis="y")

plt.tight_layout()

plt.show()
```

# Step 38 – Calculate the Correlation Matrix

## Objective

Correlation measures the strength and direction of the relationship between the daily returns of two stocks.

The correlation coefficient ranges from:

- +1 → Perfect positive correlation
- 0 → No correlation
- -1 → Perfect negative correlation

Understanding correlation helps investors build diversified portfolios by combining assets that do not move together.

```python
# ==========================================
# Calculate Correlation Matrix
# ==========================================

correlation_matrix = daily_returns.corr()

print("Correlation Matrix")

correlation_matrix
```

# Observation

The correlation matrix shows how closely the selected stocks move together.

## Business Insights

- Values close to +1 indicate that stocks tend to move in the same direction.
- Values close to 0 indicate little relationship.
- Negative values indicate that stocks tend to move in opposite directions.

Investors use correlation analysis to improve diversification and reduce overall portfolio risk.
